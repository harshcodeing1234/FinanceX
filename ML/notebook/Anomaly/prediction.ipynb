```python
import pandas as pd
import joblib
```

```python
model = joblib.load(r"C:\Users\shali\OneDrive\Desktop\FinanceX\ML\model\FinanceX_anomaly_model.pkl")
```

```python
transaction = {
    "amount": 1850,
    "category": "Shopping",
    "user_average_amount": 6000,
    "category_average_amount": 6500,
    "transaction_frequency": 2.5,
    "balance": 120000,
    "day_of_week": 5,
    "transaction_hour": 21,
    "is_weekend": 1,
    "amount_vs_user_avg": 7.5,
    "amount_vs_category_avg": 6.92,
    "recent_transaction_count": 1
}
```

```python
df = pd.DataFrame([transaction])
```

```python
prediction = model.predict(df)
print(f"Anomaly Prediction: {'Unusual transaction' if prediction[0] == -1 else 'Normal transaction'}")
```

```python
score = model.decision_function(df)[0]

print("Prediction:", prediction[0])
print("Anomaly Score:", round(score, 4))
```
