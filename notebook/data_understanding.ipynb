#### Outliers Observ

```python
for col in ["transaction_amount", "account_balance", "emi_amount"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"{col}")
    print("Lower:", lower_bound)
    print("Upper:", upper_bound)
    print("Outliers:", count)
    print()
```

```python
df[[
    "transaction_amount",
    "account_balance",
    "emi_amount"
]].describe().T
```

```python
print("Negative transaction amounts:",
      (df["transaction_amount"] < 0).sum())

print("Negative account balances:",
      (df["account_balance"] < 0).sum())

print("Negative EMI amounts:",
      (df["emi_amount"] < 0).sum())
```

```python
print("Top transaction amounts:")
print(df["transaction_amount"].nlargest(20).to_list())

print("\nTop account balances:")
print(df["account_balance"].nlargest(20).to_list())

print("\nTop EMI amounts:")
print(df["emi_amount"].nlargest(20).to_list())
```

Outlier Analysis:
- transaction_amount, account_balance and emi_amount show strong right-skewed distributions.
- IQR method identifies a large number of statistical outliers.
- These observations are not removed at this stage because extreme financial values may represent legitimate transactions.
- Further validation is required before deciding whether any values are erroneous.
- transaction_hour appears to have a valid 0–24 range.
