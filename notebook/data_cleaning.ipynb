```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.info()
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
```

```python
cat_col = df.select_dtypes(include='object').columns[1:]

for col in cat_col:
    print(df[col].value_counts())
    print("\n")
```

```python
df.groupby("transaction_status")["transaction_amount"].agg(
    ["count", "sum", "mean", "min", "max"]
)
```

```python
df['transaction_direction'].value_counts()
```

```python
pd.crosstab(
    df["transaction_status"],
    df["transaction_direction"]
)
```

```python
df.groupby("transaction_direction")["transaction_amount"].describe()
```

```python
df["transaction_hour"].value_counts().sort_index()
```
